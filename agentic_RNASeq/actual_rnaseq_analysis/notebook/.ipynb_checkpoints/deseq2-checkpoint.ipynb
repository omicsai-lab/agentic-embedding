```python
# Install PyDESeq2 if not installed
# pip install pydeseq2 pandas numpy



# 1) Read raw counts (genes x samples)
counts = pd.read_csv("counts_matrix.tsv", sep="\t", index_col=0)  # first column = gene names

# 2) Read metadata (must have a 'condition' column, even if we don't use it yet)
metadata = pd.read_csv("metadata.tsv", sep="\t")
metadata = metadata.set_index("sample")   # assumes 'sample' column matches counts columns

# Ensure order matches
metadata = metadata.loc[counts.columns]

# 3) Create DESeqDataSet (normalization only)
dds = DeseqDataSet(
    counts=counts,
    clinical=metadata,
    design_factors="condition"
)
dds.fit_size_factors()  # just calculate size factors

# 4) Get normalized counts (pseudocounts)
norm_counts = dds.get_normalized_counts()
norm_counts.to_csv("normalized_counts.tsv", sep="\t")

print("✅ Normalization complete. Normalized counts saved to 'normalized_counts.tsv'")
print(norm_counts.head())
```
